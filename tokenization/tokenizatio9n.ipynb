```python
#Get all the unicode embeddings for each string character and convert it into a list of tokens
text = "The quick brown fox jumps over the lazy dog, but did the dog ever wonder why? Innovation in artificial intelligence isn't just about code; it's about understanding human nuances. I'll be there in five minutes, don't forget to bring the documentation! We're looking at a 15% increase in efficiency across all sectors this quarter."
tokens = text.encode('UTF-8')
tokens = list(map(int, tokens))

print(tokens)
```

```python
# Detect patterns in the unicode embeddings so that we can merge them later on
def getPiars(ids):
    counts = {}
    for pair in zip(ids, ids[1:]):
        counts[pair] = counts.get(pair, 0) + 1
    return counts

pairs = getPiars(tokens)
print(pairs)
```
