```python
#Get all the unicode embeddings for each string character and convert it into a list of tokens
text = "The quick brown fox jumps over the lazy dog, but did the dog ever wonder why? Innovation in artificial intelligence isn't just about code; it's about understanding human nuances. I'll be there in five minutes, don't forget to bring the documentation! We're looking at a 15% increase in efficiency across all sectors this quarter."
tokens = text.encode('UTF-8')
tokens = list(map(int, tokens))

print(tokens)
print(len(tokens))
```

```python
# Detect patterns in the unicode embeddings so that we can merge them later on
def getPairs(ids):
	counts = {}
	for pair in zip(ids, ids[1:]):
		counts[pair] = counts.get(pair, 0) + 1
	return counts

pairs = getPairs(tokens)
print(len(pairs))
```

```python
topPair = max(pairs, key=pairs.get)
topPair
```

```python
# merge token pairs in one to form a new token
def mergePairs(ids, pair, idx):
	newids = []
	i=0
	while i < len(ids):
		if i < len(ids) - 1 and ids[i] == pair[0] and ids[i+1] == pair[1]:
			newids.append(idx)
			i+=2
		else:
			newids.append(ids[i])
			i += 1
	return newids

tokens2 = mergePairs(tokens, topPair, 169)
print(tokens2)
print(len(tokens2))

```

```python
vocabSize = 339
numMerges = vocabSize - 329
ids = list(tokens)

merges = {}
for i in range(numMerges):
	pairs = getPairs(ids)
	pair = max(pairs, key=pairs.get)
	idx = 329 + i
	print(f"merging {pair} into a new token {idx}")
	ids = mergePairs(ids, pair, idx)
	merges[pair] = idx

print(merges)
```

```python
print("tokens length: ", len(tokens))
print("ids length: ", len(ids))
print(f"compression ratio: {len(tokens) / len(ids):.2f}X")
```

```python
vocab = {idx: bytes([idx]) for idx in range(256)}
for (p0, p1), idx in merges.items():
	vocab[idx] = vocab[p0] + vocab[p1]

def decode(ids):
  tokens = b"".join(vocab[idx] for idx in ids)
  text = tokens.decode("utf-8", errors="replace") # errors="replace ensures that the bytes decoding follows the utf-8 standards"
  return text

print(decode([128]))
```

```python
def encode(text):
  tokens = list(text.encode("utf-8"))
  while len(tokens) >= 2:
    pairs = getPairs(tokens)
    pair = min(pairs, key=lambda p: merges.get(p, float("inf")))
    if pair not in merges:
      break # nothing else can be merged
    idx = merges[pair]
    tokens = mergePairs(tokens, pair, idx)
  return tokens
    
print(encode("Hello world"))
```
